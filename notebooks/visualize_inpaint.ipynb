```python
from pathlib import Path
from PIL import Image
import random
import matplotlib.pyplot as plt
```

```python
# paths
basic_path = Path('../results/hexp1.1/inpaint/interp')
deepfill_path = Path('../results/hexp1.1/inpaint/deepfill')
ddnm_100_path = Path('../results/hexp1.1/inpaint/ddnm_0.90_100')
ddnm_1000_path = Path('../results/hexp1.1/inpaint/ddnm_0.90_1000')
occ_path = Path('../data/human_occ_exp/hexp1.1')

inpaint_paths = [basic_path, deepfill_path, ddnm_100_path, ddnm_1000_path]
inpaint_names = ['Interp.', 'DeepFill', 'DDNM, T = 100', 'DDNM, T = 1000']

occ_fracs = [0.125, 0.5, 0.875]
freqs = [1, 8, 128]

random.seed(2)

nrow = len(inpaint_names) + 2
nim = len(occ_fracs) * len(freqs)
fig, axs = plt.subplots(nim, nrow, constrained_layout=True)
fig.set_size_inches(3 * nrow, 3 * nim)
col = 0
for occ_frac in occ_fracs:
    for freq in freqs:
        npix_occ = round(occ_frac * 224**2)
        # pick subexp
        subexp_name = f'simplex_noise_{freq}_{npix_occ}'
        occ_subexp_path = occ_path / subexp_name

        # pick image
        occ_im_paths = sorted((occ_subexp_path / 'images').glob('*'))
        occ_im_path = random.choice(occ_im_paths)
        im_name = occ_im_path.name
        gt_im_name = '_'.join(im_name.split('_')[:3]) + '_empty' + '.png'

        # plot
        ims = []
        names = []
        # ground truth
        gt_im_path = occ_path / 'empty' / 'images' / gt_im_name
        gt_im = Image.open(gt_im_path)
        ims.append(gt_im)
        names.append('Unoccluded')
        # occluded
        occ_im = Image.open(occ_im_path)
        ims.append(occ_im)
        names.append('Occluded')
        # inpainted
        for inpaint_path, inpaint_name in zip(inpaint_paths, inpaint_names):
            inpaint_im = Image.open(inpaint_path / subexp_name / im_name)
            ims.append(inpaint_im)
            names.append(f'{inpaint_name}')
        for row, (ax, im, name) in enumerate(zip(axs[col], ims, names)):
            ax.imshow(im)
            ax.set_xticks([])
            ax.set_yticks([])
            for side in ['left', 'right', 'top', 'bottom']:
                ax.spines[side].set_visible(False)
            if col == 0:
                ax.set_title(name, fontsize=20)
            if row == 0:
                letter = chr(ord('a') + col)
                ax.set_ylabel(f'({letter})\n\nf = {occ_frac}\n\nν = {freq}', fontsize=20, rotation=0)
                ax.yaxis.set_label_coords(-0.3, 0.2)
        col += 1
plt.savefig('out/inpaint_samp_im_comp.png')
plt.show()
```
